```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

# 1. Create DataFrame with random numbers
np.random.seed(42)
df = pd.DataFrame(np.random.randn(50, 4), columns=["f1", "f2", "f3", "f4"])

# 2. Scatter plot of f2 vs f3
plt.scatter(df["f2"], df["f3"])
plt.xlabel("f2")
plt.ylabel("f3")
plt.title("f2 vs f3 Scatter Plot")
plt.savefig("scatter.png")
plt.show()

# 3. Row-wise mean into column f5
df["f5"] = df.mean(axis=1)

# 4. Save to Excel (.xlsx)
df.to_excel("output.xlsx", index=False)
```
